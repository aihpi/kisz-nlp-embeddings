In this notebook, we'll explore **Doc2Vec**, a model for learning document-level embeddings developed by Le and Mikolov (paper [here](https://cs.stanford.edu/~quocle/paragraph_vector.pdf)). Unlike word embeddings like Word2Vec or FastText, Doc2Vec represents entire documents as dense vectors, which can be useful for tasks such as document similarity and classification.

<h3>1. Overview
</div>

Doc2Vec is a neural embedding model that learns fixed-length vector representations for documents, enabling semantic comparisons between texts of varying length. It extends Word2Vec by associating each document with a unique vector trained to represent the document’s meaning, making it ideal for tasks like document similarity, classification, clustering, and semantic search.

Doc2Vec comes in two main variants:

- **Distributed Memory (DM)**

    Predicts a target word using the surrounding context and the document vector.

    The document vector acts as a memory that helps predict content based on the document’s semantics.

    Preserves word order and captures the topic of the document.

    Generally provides richer representations.

- **Distributed Bag of Words (DBOW)**

    Ignores word order and context; predicts words directly from the document vector.

    Faster and more memory-efficient.

    Often performs best when combined with DM (e.g., by concatenating their vectors).

    Can optionally train word vectors too (dbow_words=True).

Together, these two training modes allow Doc2Vec to learn robust document embeddings that generalize well across a wide range of NLP applications.

<h3>5. Advantages and disadvantages of Doc2Vec
</div>

Let's summarize the pros and cons of FastText embeddings, and see where can they be used.
        
#### Advantages:

- **Document-Level Embeddings**: Doc2Vec extends Word2Vec by generating vector representations not just for words but for entire documents, sentences, or paragraphs. This allows for capturing the semantics of larger text units in a fixed-length vector, making it ideal for document-level tasks.
- **Handles Variable-Length Texts**: Doc2Vec can process inputs of varying lengths—sentences, paragraphs, or full documents—into a consistent vector size, making it useful in many real-world NLP applications.
- **Semantic Similarity Beyond Words**: By embedding documents in the same space as words, Doc2Vec allows for similarity comparisons between documents, as well as between words and documents, improving semantic search and clustering.

#### Disadvantages:

- **Training Complexity**: Training Doc2Vec is generally more complex and computationally intensive than Word2Vec or FastText, especially for large corpora. It requires maintaining and updating document vectors during training, increasing memory usage.
- **Slower Convergence**: Compared to Word2Vec, Doc2Vec typically takes longer to train and may be more sensitive to hyperparameters, requiring careful tuning to achieve good performance.
- **Static Representations**: Like Word2Vec and FastText, Doc2Vec produces static embeddings—each document has one fixed vector, regardless of usage or changing context.
- **Dependency on Corpus**: The quality of the document embeddings heavily depends on the quality and coverage of the training corpus. Out-of-domain or new documents may not be well-represented without retraining or inference.

#### Applications:

- **Text Classification**: Doc2Vec vectors serve as strong features for categorizing documents by topic, sentiment, or intent across tasks like email classification, news categorization, and more.
- **Spell Correction & Autocomplete**: While not as specialized for subword tasks as FastText, Doc2Vec can contribute to context-aware spell correction and completion at the document level when combined with other models.
- **Search and Information Retrieval**: Doc2Vec enhances semantic search by embedding documents and queries in the same vector space, enabling retrieval based on meaning rather than keyword matching.
- **Low-Resource Language NLP**: When trained on relatively small but domain-specific corpora, Doc2Vec can still generate meaningful document embeddings, making it useful in low-resource settings.
- **Multilingual Embedding Alignment**: Doc2Vec has been adapted for multilingual or cross-lingual setups, enabling similarity comparisons and translation tasks across languages by aligning document embeddings.
